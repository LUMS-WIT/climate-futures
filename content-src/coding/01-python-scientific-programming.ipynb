A **variable** stores a value under a name, using `=`. A good variable name says what the value is, and a comment states the unit. Python does not track units, so the programmer must.

```python
solar_constant = 1361        # W/m², total solar irradiance at the top of the atmosphere
albedo = 0.3                 # dimensionless, fraction of sunlight reflected by the Earth
sigma = 5.670374419e-8       # W/m²/K⁴, Stefan–Boltzmann constant (e-8 means × 10⁻⁸)

absorbed = solar_constant * (1 - albedo) / 4   # W/m², averaged over the whole sphere
print(absorbed)
```

where $S_0$ is the solar constant, $\alpha$ the albedo, and $\sigma$ the Stefan–Boltzmann constant.

## 4 · Lists and loops

A **list** holds several values in order, written between square brackets. Items are accessed by their **index**, which starts at **zero**. A negative index counts from the end.

```python
years = [2020, 2030, 2040, 2050]
print(years[0])    # the first item
print(years[-1])   # the last item
print(len(years))  # the number of items
```

A **`for` loop** repeats a block of code once for each item of a list. The indented lines form the body of the loop; indentation is part of Python's syntax, not a matter of style.

`range(n)` produces the numbers 0, 1, ..., n − 1, and `.append()` adds an item to the end of a list.

```python
for y in years:
    print(f"Year {y} is {y - 2020} years after 2020")

squares = []
for i in range(5):
    squares.append(i ** 2)
print(squares)
```

## 6 · Functions

## 10 · A small model: the remaining carbon budget

Lecture 4 introduced the near-linear relationship between cumulative CO₂ emissions and warming, which gives rise to the idea of a **remaining carbon budget**. The model below asks a simple question: if emissions decline in a straight line from today's level to zero, how many years can that decline take before the budget is used up?

The two inputs are round values for illustration: a remaining budget of 500 Gt CO₂ and current emissions of 40 Gt CO₂/yr. Exercise 17 asks for a sensitivity analysis, so the exact values matter less than the method.

```python
def cumulative_emissions(start_emissions, years_to_zero):
    """Total emissions (Gt CO2) if annual emissions fall linearly to zero over `years_to_zero` years."""
    t = np.arange(years_to_zero + 1)                     # 0, 1, ..., years_to_zero
    annual = start_emissions * (1 - t / years_to_zero)   # straight line to zero
    return annual.sum()

print(cumulative_emissions(40, 30), "Gt CO2 if emissions reach zero after 30 years")
```

## 11 · Reading error messages

Errors are a normal part of programming. Python reports an error with a **traceback**; the most useful information is usually on the **last line**, which names the type of error and gives a short explanation. Common types:

| Error | Typical cause |
|---|---|
| `NameError` | A variable is misspelt, or the cell that defines it has not been run. |
| `TypeError` | An operation is applied to the wrong type, for example adding text to a number. |
| `IndexError` | A list position does not exist, for example the fifth item of a four-item list. |
| `SyntaxError` | A bracket, quotation mark, or colon is missing. |
| `KeyError` | A column or dictionary key does not exist in pandas. |

## 12 · Good practice for reproducible work

Lecture 9 introduced the FAIR principles and version control. In a notebook, reproducibility depends on a few habits:

1. **Run from top to bottom.** Before submitting, use **Runtime → Restart and run all** (Colab) or **Kernel → Restart & Run All** (Jupyter). A notebook that only works when its cells are run in a particular unusual order is not reproducible.
2. **Name things clearly** and **state units** in names or comments.
3. **Do not repeat code.** When the same calculation appears twice, write a function.
4. **Record the source of every dataset**, including the version and the date of access.
5. **Keep inputs separate from calculations,** so that assumptions (such as the 500 Gt CO₂ budget above) are defined once, near the top, and are easy to change.
